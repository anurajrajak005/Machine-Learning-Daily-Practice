```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
```

```python
X,y=make_regression(n_samples=4,n_features=1,n_targets=1, random_state=34,noise=150)
print(X)
print(y)
```

```python
plt.scatter(X,y)
```

```python
Lr=LinearRegression()
Lr.fit(X,y)
```

```python
print("coef=",Lr.coef_)
print("intercept",Lr.intercept_)
y_pred=Lr.predict(X)
print('Y_pred=',y_pred)
print("Actual_pred",y)
```

```python
plt.scatter(X,y,c="red")
plt.plot(X,y_pred)
```

```python
# Slope=Square(b):After differnetiate=-2*sum(y-mx-b)
b=0
m=-66.12
loss_slope=-2*np.sum(y-m*X.ravel()-b)
print(loss_slope)
```

```python
# Iteration 1
#lets take Learning rate=0.1
lr=0.1
step_size=lr*loss_slope
print(step_size)
```

```python
# Calculate new intercept
b=b-step_size
print(b)
```

```python
y_pred1=((-66.12*X)+b).reshape(4)
plt.scatter(X,y)
# plt.plot(X,Lr.predict(X),color="red",label="OLS")
plt.plot(X,y_pred1,color="blue",label='b={}'.format(b))
plt.plot(X,y_pred,color="green",label="OLS")
plt.legend()
plt.show()
```
