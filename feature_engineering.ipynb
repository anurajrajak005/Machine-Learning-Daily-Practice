```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
df = pd.DataFrame(
    {
        'Country': ['India', 'USA', 'India', 'UK', np.nan, 'India', 'UK'],
        'Purchased': [1,0,1,1,np.nan,0,1],
        'Turnover': [50000, np.nan, 80000, 70000, 45000, 65000, 44000]
    }
)
```

```python
sm = SimpleImputer(strategy='most_frequent')
```

```python
df[['Country']] = sm.fit_transform(df[['Country']])
```

```python
df
```

```python
sm = SimpleImputer(fill_value=5000 ,strategy='constant')
df[['Turnover']] = sm.fit_transform(df[['Turnover']])
```
