##### Label Encoder 

```python
L= LabelEncoder()
dt['gender']=L.fit_transform(dt['gender'])
dt['gender']
```

```python
ords=OrdinalEncoder(categories=[['sourcing','referred','other']])
dt['recruitment_channel']=ords.fit_transform(dt[['recruitment_channel']])
```

```python
dt
```

- Frequencies Encoder 

```python
Tr=pd.read_csv('Promotions.csv')
dp = Tr.department.value_counts().to_dict()
dp()
```

```python
Tr['department']=Tr['department'].map(dp)
Tr
```

- Target Encoder 

```python
dfs = pd.DataFrame({'country':['india','india','usa','usa','japan'],
                    'purchased':[100,200,300,400,500]})
```

```python
dfs
```

```python
# tgs = TargetEncoder(smooth=5, target_type='continuous')
```

```python
tg=TargetEncoder(smooth=1,target_type='continuous')
tg
```

```python
tg.fit(dfs[['country']],dfs['purchased'])
```

```python
dfs['country']=tg.transform(dfs[['country']])
```
