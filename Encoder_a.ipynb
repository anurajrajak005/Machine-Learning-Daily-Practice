```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'color':['Red','Blue','Green','Blue','Red'],
    'shape':['circle','Square','Triangle','Circle','Square']
})
```

```python
print("original Data")
print(df)
```

```python
le_color = LabelEncoder()
le_shape = LabelEncoder()
```

```python
df['color_encoded']= le_color.fit_transform(df['color'])
df['shape_encoded']= le_shape.fit_transform(df['shape'])
```

```python
print("\n One-Hot Encoded Data:")
print("df_encoded_pandas")
df
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.DataFrame({
    'color':['Red','Blue','Green','Blue','Red'],
    'shape':['circle','Square','Triangle','Circle','Square']
})
print(df)
```

```python
encoder=OneHotEncoder(sparse_output=False)
encoded=encoder.fit_transform(df[["Color","Shape"]])
df_encoded=pd.DataFrame(encoded,columns=encoder.get_feature_names_out(["Color","Shape"]))
print(df_encoded)
```

```python
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    'size':['Small','Medium','Large','Medium','Small'],
    'Education':['Primary','Secondary','Graduate','PostGraduate','Graduate']
})
print(df)
```

```python
print("Orginial Data")
print(df)
```

```python
# order def 
size_order = ['Small','Medium','Large']
edu_order = ['Primary','Secondary','Graduate','PostGraduate']
```

```python
encoder = OrdinalEncoder(categories = [size_order,edu_order])
df_encoded= df.copy()
df_encoded[['size_encoded','Edu_encoded']]=encoder.fit_transform(df[["size","Education"]])

print(df_encoded)
```
