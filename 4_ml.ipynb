```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,LabelEncoder,OrdinalEncoder,TargetEncoder,StandardScaler,MinMaxScaler
from sklearn.impute import SimpleImputer
```

```python
df=pd.read_csv("C:/Users/Gyanp/Downloads/PROPERTIES.csv")
```

```python
df
```

```python
tr = ColumnTransformer(transformers = [
    ('ohe', OneHotEncoder(drop = 'first', sparse_output = False), ['Area']),
    ('sq', SimpleImputer(strategy = 'mean'), ['Sqft']),
    ('rv', OrdinalEncoder(categories = [['Poor', 'Average', 'Good']]), ['Review'])
], remainder = 'passthrough')
```

```python
k = tr.fit_transform(df)
print(k)
j = pd.DataFrame(k, columns = ['Area1', 'Area2', 'Sqft', 'Review', 'Price'])
j = j.map(int)
print(j)
```

```python
# Feature Extraction
j['Mobile'] = 10
print(j)
```
