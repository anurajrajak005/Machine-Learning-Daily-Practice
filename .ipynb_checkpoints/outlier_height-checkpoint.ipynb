```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
data=pd.read_csv("outliee_height.csv")
data
```

```python
sns.boxplot(data,x=data['HEIGHT'])
plt.show()
```

```python
Q1=data["HEIGHT"].quantile(.25)
print(Q1)
```

```python
Q2=data["HEIGHT"].quantile(.50)
print(Q2)
```

```python
Q3=data["HEIGHT"].quantile(.75)
print(Q3)
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
LB=Q1-1.5*IQR
print('LBO', LB)
```

```python
UB=Q3+1.5*IQR
print('UBO', UB)
```

```python
j=data[(data['HEIGHT']<LB) | (data["HEIGHT"]>UB)]
print(j)
```

```python
newdata=data[(data['HEIGHT']>=LB) & (data["HEIGHT"]<=UB)]
print(newdata)
```

```python
sns.boxplot(newdata,x=newdata['HEIGHT'])
plt.show()
```

```python
n=data[(data['HEIGHT']>=LB) & (data["HEIGHT"]<=UB-2)]
print(n)
```

```python
sns.boxplot(n,x=n['HEIGHT'])
plt.show()
```
