```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import OneHotEncoder,LabelEncoder,OrdinalEncoder, TargetEncoder
```

```python
df = pd.DataFrame(
    {
        'Country': ['India', 'USA', 'India', 'UK', 'USA', 'India', 'UK','USA'],
        'Purchased': [1,0,1,1,0,0,1,1]
    }
)
```

```python
df
```

```python
data = df
data
```

#### using pandas

```python
k = data.groupby('Country')['Purchased'].mean()
print(k)
```

```python
data['Country'] = data['Country'].map(k)
```

```python
data
```

### using targetEncoder

```python
Tr = TargetEncoder()
```

```python
# Tr
```

```python
l = Tr.fit(data[["Country"]],data["Purchased"])
l
```

```python
Tr.transform([[l]])
```
