```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import OneHotEncoder,LabelEncoder,OrdinalEncoder,TargetEncoder,StandardScaler,MinMaxScaler
from sklearn.impute import SimpleImputer
```

```python
df=pd.read_csv("Kmeanssalry - Kmeanssalry.csv")
df
```

```python
df1=df.copy()
df1.head(2)
```

```python
fig = plt.figure(figsize=(8,5))
ax = fig.add_subplot(1,1,1)

df['Age'].plot(kind='kde')
df['Income'].plot(kind='kde')
plt.show()
```

## MinMaxScaler

```python
df1.head(2)
```

```python
m=MinMaxScaler()
df1['Age']=m.fit_transform(df1[['Age']])
df1['Income']=m.fit_transform(df1[['Income']])
df1.head(2)
```

```python
#df['Age'].describe()
#df['Income'].describe()
```

```python
fig = plt.figure(figsize=(8,5))
ax = fig.add_subplot(1,1,1)

df1['Age'].plot(kind='kde')
df1['Income'].plot(kind='kde')
plt.show()
```

## StandardScaler (Standardization)

```python
df2=df.copy()
sc = StandardScaler()

df2["Age"] = sc.fit_transform(df2[["Age"]])
df2['Income']=sc.fit_transform(df2[['Income']])
df2
```

```python
fig = plt.figure(figsize=(8,5))
ax = fig.add_subplot(1,1,1)

df2['Age'].plot(kind='kde')
df2['Income'].plot(kind='kde')
plt.show()
```
