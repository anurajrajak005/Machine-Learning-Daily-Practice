```python

import numpy as np
import matplotlib.pyplot as plt

class GradientDescent:

    def __init__(self, m, b, lr, epochs):
        self.m = m
        self.b = b
        self.lr = lr
        self.epochs = epochs

    def fit(self, X, y):

        X = X.ravel()

        for i in range(self.epochs):

            # Calculate gradient of b
            loss_slope = -2 * np.sum(
                y - self.m * X - self.b
            )

            # Update b
            self.b = self.b - (self.lr * loss_slope)

            # Prediction
            y_pred = self.m * X + self.b

            # Plot each iteration
            plt.plot(X, y_pred)

            print("Iteration:", i+1)
            print("Slope:", self.m)
            print("Intercept:", self.b)

        plt.scatter(X, y, color="red", label="Actual Data")
        
        plt.show()

        return self


# Create object
gd = GradientDescent(
    m=-66.12677565,
    b=-150,
    lr=0.01,
    epochs=15
)

# Train model
gd.fit(X, y)
```

```python

import numpy as np

class mygd:

    def __init__(self, learningrate=0.01, epochs=100):
        self.coef_ = None
        self.intercept_ = None
        self.lr = learningrate
        self.epochs = epochs

    def fit(self, xtrain, ytrain):

        self.intercept_ = 0
        self.coef_ = np.ones(xtrain.shape[1])

        for i in range(self.epochs):

            # Prediction
            y_hat = np.dot(xtrain, self.coef_) + self.intercept_

            # Intercept gradient
            intercept_der = -2 * np.mean(ytrain - y_hat)

            # Update intercept
            self.intercept_ = self.intercept_ - (
                self.lr * intercept_der
            )

            # Coefficient gradient
            coef_der = -2 * np.dot(
                (ytrain - y_hat), xtrain
            ) / xtrain.shape[0]

            # Update coefficients
            self.coef_ = self.coef_ - (
                self.lr * coef_der
            )

        print("Intercept:", self.intercept_)
        print("Coefficients:", self.coef_)

    def predict(self, xtest):

        return np.dot(xtest, self.coef_) + self.intercept_


# Create object
gd = mygd(learningrate=0.1, epochs=1200)

# Train model
gd.fit(X, y)

# Predictions
y_pred = gd.predict(X)

print("Predictions:", y_pred)
```

```python
class mbmygd:
    def __init__(self,batchsize,learningrate=0.01,epochs=100):
        self.coef_=None
        self.intercept_=None
        self.lr=learningrate
        self.epochs=epochs
        self.batch_size=batchsize
    def fit(self,xtrain,ytrain):
        self.intercept_=0
        self.coef_=np.ones(xtrain.shape[1])
        for i in range(self.epochs):
            for j in range(int(xtrain.shape[0]/self.batch_size)):
                idx=random.sample(range(xtrain.shape[0]),self.batch_size)
                y_hat=np.dot(xtrain[idx],self.coef_)+(self.intercept_)
                intercept_der=-2*np.mean(ytrain[idx]-y_hat)
                self.intercept_=self.intercept_-(self.lr*intercept_der)
                coef_der=-2*np.dot((ytrain[idx]-y_hat),xtrain[idx])
                self.coef_=self.coef_-(self.lr*coef_der)
                print(self.intercept_,self.coef_)
    def predict(self,xtest): 
        return np.dot(xtest,self.coef_)+self.intercept_
```

```python
# mini batchgradient ddescent 
import numpy as np
import matplotlib.pyplot as plt 
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression 
from sklearn.metrics import r2_score
import random

X,y=make_regression(n_samples=353,n_features=3,n_targets=1,noise=80,random_state=13)
# print(X)
print(X.shape)
# print(y)

lr=LinearRegression()
lr.fit(X,y)

class mbmyggd:
    def __init__(self,batchsize,learningrate=0.01,epochs=100):
        self.coef_=None
        self.interceft_=None
        self.lr=learningrate
        self.epochs=epochs
        self.batch_size=batchsize

    def fit (self,xtrain,ytrain):
        self.interceft_=0
        self.coef_=np.ones(xtrain.shape[1])
        for i in range(self.epochs):
            for j in range(int(xtrain.shape[0]/self.batch_size)):
                idx=random.sample(range(xtrain.shape[0]),self.batch_size)
                y_hat=np.dot(xtrain[idx],self.coef_)+(self.interceft_)
                interceft_der=-2*np.mean(ytrain[idx]-y_hat)
                self.interceft_=self.interceft_-(self.lr*interceft_der)

                coef_der=-2*np.dot((ytrain[idx]-y_hat),xtrain[idx])
                self.coef_=self.coef_-(self.lr*coef_der)
        print(self.interceft_,self.coef_)

    def predict (self,xtest):
        return np.dot(xtest,self.coef_)+(self.interceft_)

```

```python
m=mbmyggd(batchsize=int(X.shape[0]/10),learningrate=0.01,epochs=100)
```

```python
m.fit(X,y)
```

```python
y_pred=m.predict(X)
print(r2_score(y,y_pred))
```
