```python
dt = {
    "Pizza": np.arange(8,41,2),
    "Price": np.arange(10,59,3)
}
```

```python
dt
```

```python
df = pd.DataFrame(dt)
```

```python
df.head()
```

```python
X = df[['Pizza']]
y = df['Price']
```

```python
Xtrain, Xtest, ytrain, ytest = train_test_split(X, y, test_size=0.2)
```

```python
Xtrain
```

```python
Xtest
```

```python
lr = LinearRegression()
lr.fit(Xtrain, ytrain)
```

```python
lr.predict([[10]])
```

```python
lr.intercept_
```

```python
lr.coef_
```
